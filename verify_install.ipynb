# Check your installation

Run the cell below before the course. It shows which Python your kernel is using
and imports the packages the notebooks need.

If something fails, the usual causes are that the requirements were not installed
(`pip install -r requirements.txt`) or that the notebook was opened with the wrong
kernel. The Python path printed below tells you which environment you are in.


```python
import importlib
import sys

print("Python version:")
print(sys.version)

print("\nPython executable:")
print(sys.executable)
print()

packages = [
    "numpy",
    "sklearn",
    "matplotlib",
    "seaborn",
    "PIL",
    "requests",
    "torch",
    "torchinfo",
    "transformers",
    "datasets",
    "evaluate",
]

missing = []
for package in packages:
    try:
        importlib.import_module(package)
        print(f"\u2713 {package}")
    except Exception as error:
        missing.append(package)
        print(f"\u2717 {package}: {error}")

if missing:
    print("\n\u274c Could not import: " + ", ".join(missing))
    print("Install the requirements and check that the kernel above is the one you created.")
else:
    print("\n\u2705 Installation looks good. You are ready for the course.")
```
